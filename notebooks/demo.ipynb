# Simulate closed-loop
After implementing your control functionality, you can simulate the closed-loop with code that looks something like this...

```python
from pathlib import Path
import sys

import numpy as np

project_root = next(
    path for path in (Path.cwd(), Path.cwd().parent)
    if (path / "uuv_mission").is_dir()
)
sys.path.insert(0, str(project_root))

from uuv_mission.controller import PDController
from uuv_mission.dynamic import ClosedLoop, Mission, Submarine

mission = Mission.from_csv(str(project_root / "data" / "mission.csv"))
controller = PDController(kp=0.05, kd=0.1)
simulation = ClosedLoop(Submarine(), controller)
trajectory = simulation.simulate(mission, np.zeros(len(mission.reference)))

positions = trajectory.position
assert positions.shape == (len(mission.reference), 2)
assert np.isfinite(positions).all()
assert np.any(positions[:, 1] != 0), "Controller actions did not move the submarine vertically"

print(f"Integration check passed: {len(positions)} finite positions; controller moved the submarine.")
```
